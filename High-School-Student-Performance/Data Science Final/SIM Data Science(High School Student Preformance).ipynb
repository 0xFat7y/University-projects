```python
import pandas as pd

def handle_outliers(df, columns):
    for col in columns:

        if col not in df.columns:
            continue

        if not pd.api.types.is_numeric_dtype(df[col]):
            continue

        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1

        if IQR == 0:
            continue

        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR

        df[col] = df[col].clip(lower, upper)

    return df
```

### Create the following :

#### 2)Pass/Fail indicator
#### 3)Average Grade





```python
df['cgpa'] = df[['grade_1', 'grade_2', 'final_grade']].mean(axis=1)
```

```python
def grade_category(grade):
    if grade >= 18: return 'A',
    elif grade >= 15: return 'B',
    elif grade >= 12: return 'C',
    elif grade >= 10: return 'D',
    else: return 'F'
```

```python
df['grade_category'] = df['final_grade'].apply(grade_category)
```

```python
df['pass_fail'] = df['final_grade'].apply(lambda x: 'Pass' if x >= 10 else 'Fail')
```
